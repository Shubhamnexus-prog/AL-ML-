**Set Operation**

**union**

```python
A={1,2,3,4,5}
B={4,5,6,7,8}
C={8,9,10}
A|B|C
```

```python
A.union(B)
```

```python
A.union(B,C)
```

```python
A.update(B,C)
A
```

**Intersection**

```python
A={1,2,3,4,5}
B={4,5,6,7,8}
A&B
```

```python
A.intersection(B)
```

```python
A.intersection_update(B )
A
```

**Difference**

```python
A={1,2,3,4,5}
B={4,5,6,7,8}
A-B
```

```python
A.difference(B)
```

```python
B-A
```

```python
B.difference(A)
```

```python
B.difference_update(A)
B
```

**Symmetric Difference**

```python
A={1,2,3,4,5}
B={4,5,6,7,8}
A^B
```

```python
A.symmetric_difference(B)
```

```python
A.symmetric_difference_update(B)
A
```
