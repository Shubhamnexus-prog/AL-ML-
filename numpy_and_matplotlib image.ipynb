```python
#add matrix (3x4)-----


a=np.array([
    [1,2,3,4],
    [5,6,7,8],
    [3,6,7,9]
])
b=np.array([
    [2,3,4,5],
    [4,6,7,8],
    [2,3,7,8]
])
c=a+b
print( " add matrix:3x4\n"
       ,c)

           
```

```python
#diff matrix (3x4)-----


a=np.array([
    [1,2,3,4],
    [5,6,7,8],
    [3,6,7,9]
])
b=np.array([
    [2,3,4,5],
    [4,6,7,8],
    [2,3,7,8]
])
c=a-b
print( "diff matrix:3x4\n"
       ,c)
```

```python
#product matrix (3x4)-----


a=np.array([
    [1,2,3,4],
    [5,6,7,8],
    [3,6,7,9]
])
b=np.array([
    [2,3,4,5],
    [4,6,7,8],
    [2,3,7,8]
])
c=a*b
print( "product matrix:3x4\n"
       ,c)
```
