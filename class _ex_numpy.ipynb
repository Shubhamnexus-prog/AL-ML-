**Array manipulation funcation**

```python
a1=np.array([1,2,3])
reshaped=np.reshape(a1,(3,1))
print("Reshaped array:",reshaped)
```

```python
f1=np.array([[1,2],[3,4]])
flattened=np.ravel(f1)
print("flattened array:",flattened)
```

```python
e1=np.array([[1,2],[3,4]])
transposed =np.transpose(e1)
print("Transposed array:\n",transposed)
```

```python
a2=np.array([1,2])
b2=np.array([3,4])
stacked=np.vstack([a2,b2])
print("Stacked arrays:\n",stacked)
```

```python
a2=np.array([1,2])
b2=np.array([3,4])
stacked=np.hstack([a2,b2])
print("stacked arrays:\n",stacked)
```

```python
g=np.array([1,2,3,4])
added=np.add(g,2)
print("Added 2 to g:",added)
```

```python
squared=np.power(g,2)
print("Squared g:",squared)
```

```python
sqrt_val=np.sqrt(g)
print("Square root of g:",sqrt_val)
```

**statistical functions**

```python
s=np.array([1,2,3,4])
mean=np.mean(s)
print("Mean of s:",mean)
```

```python
s=np.array([1,2,3,4])
mean=np.var(s)
print("Mean of s:",mean)
```

```python
std_dev=np.std(s)
print("Standard deviation of s:",std_dev)
```

```python
minimum=np.min(s)
print("min of s:",minimum)
```

```python
maximum=np.max(s)
print("Max of s:",maximum)
```
