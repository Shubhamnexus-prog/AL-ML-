```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)
plt.title("Simple Line Plot")
plt.xlabel("X-axis")  
plt.ylabel("Y-axis")  
plt.show()
```

**Scatter plot**

```python
plt.bar(x,y)
plt.title("Bar chart")
plt.show()
```

```python


x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.scatter(x, y, color='green')
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")  
plt.show()
```

```python

categories = ['A', 'B', 'C', 'D']
values = [5, 7, 3, 8]
plt.barh(categories, values, color='purple')
plt.title("Horizontal Bar Chart")
plt.xlabel("Values") 
plt.ylabel("Categories")
plt.show()

```

```python
data = [1,2,2,3,3,3,4,4,4,4,5,5,6]

plt.hist(data, bins=6, color='skyblue', edgecolor='black')
plt.title("Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
sizes = [30, 20, 25, 25]
labels = ['A', 'B', 'C', 'D']
colors = ['gold', 'lightblue', 'lightgreen', 'pink']

plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
plt.title("Pie Chart")
plt.show()
```

```python
plt.fill_between(x, y, color="skyblue", alpha=0.5)
plt.plot(x, y, color="Slateblue", alpha=0.8)
plt.title("Area Chart")
plt.show()
```

```python
days = [1, 2, 3, 4, 5]
sleep = [7, 6, 8, 5, 7]
work = [8, 9, 7, 8, 8]
exercise = [1, 1, 1, 2, 1]

plt.stackplot(days, sleep, work, exercise, labels=['Sleep','Work','Exercise'], colors=['skyblue','orange','green'])
plt.legend(loc='upper left')
plt.title("Stack Plot")
plt.show()
```

```python
data = [7, 8, 5, 6, 9, 10, 6, 7]

plt.boxplot(data)
plt.title("Box Plot")
plt.show()
```

```python
plt.plot(x, y, label="Line")
plt.scatter(x, y, color="red", label="Points")
plt.bar(categories, values, alpha=0.5, label="Bars")
plt.legend()
plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

x = [1, 2, 3, 4, 5]
y = [2, 3, 4, 5, 6]
z = [5, 6, 2, 3, 4]

ax.plot3D(x, y, z, color='red')
ax.set_title("3D Line Plot")
plt.show()
```
